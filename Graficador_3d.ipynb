```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Definir la función de la pizarra (asumiendo C = 1 para ver la forma)
def target_function(x1, x2):
    exponent = -( (x1**2)*(x2**2) + x1**2 + x2**2 - 8*x1 - 8*x2 ) / 2
    return np.exp(exponent)

# 2. Crear la malla de puntos para x1 y x2
# El rango de -0.5 a 4.5 es ideal para capturar los picos de esta función
x1 = np.linspace(-0.5, 4.5, 150)
x2 = np.linspace(-0.5, 4.5, 150)
X1, X2 = np.meshgrid(x1, x2)
Z = target_function(X1, X2)

# 3. Configurar la figura y los ejes 3D
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(projection='3d')

# 4. Graficar la superficie
surf = ax.plot_surface(X1, X2, Z, cmap='viridis', edgecolor='none', alpha=0.9)

# 5. Personalizar etiquetas y vista
ax.set_title('Gráfica 3D de la función objetivo', fontsize=14)
ax.set_xlabel('$x_1$', fontsize=12)
ax.set_ylabel('$x_2$', fontsize=12)
ax.set_zlabel('$f(x)$', fontsize=12)

# Añadir barra de colores lateral
fig.colorbar(surf, shrink=0.5, aspect=5, label='Valor de la función')

# Mostrar la gráfica interactiva
plt.show()
```
