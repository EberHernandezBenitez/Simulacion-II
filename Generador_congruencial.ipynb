Para el generador congruencial
$$x_{i+1}= ax_i+b (mod m)$$

```python
def generador(a,b,m,x0):
  x1=(a*x0+b) %m
  return x1
```

```python
a= 11
b= 5
m= 16
x0= 1
n= 10
muestra=[]
for i in range(n):
  nuevo= generador(a,b,m,x0)
  muestra.append(nuevo)
  x0=nuevo/m
print(muestra)
```
