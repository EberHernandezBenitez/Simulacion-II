# Agregamos la matriz de transición.

```python
P=Matrix([[0.6,0.4],[0.2,0.8]])
```

Dada la matriz de probabilidades de transición $P$ y el estado inicial $\pi_0$ de una cadena de Markov, la distribución en el tiempo siguiente se calcula como:
$$\pi_{n+1}= \pi_n P$$

```python
p=Matrix([[1,0]])
```

```python
p
```

Para calcular la probabilidad de que llueva el el sábado:

```python
sabado=p*P**5
```

```python
sabado
```

¿Cuál es la probabilidad de que a largo plazo un día sea soleado?
Acompañe su respuesta con una gráfica.

```python
def dias(P,p,n):
  matrices=[]
  for i in range(n):
    ojo=p*P**i
    matrices.append(ojo[1])
  return matrices
```

```python
grafica=dias(P,p,60)
plt.plot(grafica)
plt.title("Probabilidad de que en el día n este el solecito")
plt.xlabel("Días(n)")
plt.ylabel("Probabilidad")
```

Por lo tanto, la probabilidad es aproximadamente 0.6666
