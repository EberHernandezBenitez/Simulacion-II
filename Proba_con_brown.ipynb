# Ejemplo:
Encuentre la probabilidad de que el MB con parámetro de $\mu$=0.6 y $\sigma^2=0.25$, tome valores entre 1 y 3 al tiempo $t=4$.

# Solución analítica:
$X_t=0.6t+0.5B_t$, sustituyendo t=4 tenemos las siguiente expresión:
$$P(1<X_4<3)=P(1<2.4+\frac{1}{2}B_4<3)$$
$$=P(-1.4<\frac{1}{2}B_4<0.6)$$
$$=P(-2.8<B_4<1.2)$$
$$=\int_{-2.8}^{1.2} \frac{1}{\sqrt{2*4*π}}e^{-\frac{x^2}{8}}dx$$

$$=0.645$$


# Solución por simulación:

Para ello, usaremos puentes browneanos con deriva:

```python
import numpy as np
import random as r
import matplotlib.pyplot as plt
```

```python
def brownie(T,L):
    veo = []
    w_0 = 0
    dt = T/L
    for _ in range(1,L):
        dw = r.gauss(0,1)
        w_1 = w_0 + np.sqrt(dt)*dw
        veo.append(w_1)

        w_0 = w_1
    veo.insert(0,0)
    return veo
```

```python
def deriva(mu, sigma, T, n):
  A = np.array(brownie(T, n))
  t = np.linspace(0, T, n)

  veo = mu * t + sigma * A

  return veo.tolist()
```

```python
def prob(mu, sigma, T, n, N):
  conteo = 0

  for _ in range(N):
    veo = deriva(mu, sigma, T, n)
    valor_final = veo[-1]  # X_t en el tiempo T = 4
    if 1 <= valor_final <= 3:
      conteo += 1

  # Calculamos la probabilidad
  probabilidad = conteo / N
  return probabilidad
```
